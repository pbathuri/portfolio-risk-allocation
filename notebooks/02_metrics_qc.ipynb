## Week 2: return and risk
For equal weights and covariance [[.04,.01],[.01,.09]], calculate variance by hand, including cross terms.

```python
import numpy as np
w=np.array([.5,.5]); cov=np.array([[.04,.01],[.01,.09]])
print("Variance",w@cov@w,"Volatility",np.sqrt(w@cov@w))
train,valid,test=p.split(DATA)
print(train.mean()*12)
train.corr()
```

```python
import matplotlib.pyplot as plt
fig,ax=plt.subplots(figsize=(6,5))
im=ax.imshow(train.corr(),vmin=-1,vmax=1,cmap="RdBu_r")
ax.set_xticks(range(train.shape[1]),train.columns,rotation=90)
ax.set_yticks(range(train.shape[1]),train.columns)
fig.colorbar(im,ax=ax);fig.tight_layout();fig.savefig(OUT/"correlation.png",dpi=150)
plt.close(fig)
from IPython.display import display, Image
display(Image(filename=str(OUT/"correlation.png")))
```
